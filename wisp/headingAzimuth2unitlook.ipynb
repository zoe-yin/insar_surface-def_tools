```python
import numpy as np

def sat_to_ground_look_vector(heading_deg, incidence_deg, look_direction="right"):
    """
    Calculate satellite-to-ground satellite look unit vector in ENU coordinates.

    Parameters
    ----------
    heading_deg : float
        Satellite heading, clockwise from north (degrees).
    incidence_deg : float
        Radar incidence angle from vertical (degrees).
    look_direction : str
        "right" or "left" looking SAR.

    Returns
    -------
    np.ndarray
        [east, north, up] satellite-to-ground unit look vector (WISP convention).
    """

    heading = np.deg2rad(heading_deg)
    incidence = np.deg2rad(incidence_deg)

    if look_direction.lower() == "right":
        azimuth = heading + np.deg2rad(90)

    elif look_direction.lower() == "left":
        azimuth = heading - np.deg2rad(90)

    else:
        raise ValueError("look_direction must be 'right' or 'left'")

    east = np.sin(incidence) * np.sin(azimuth)
    north = np.sin(incidence) * np.cos(azimuth)
    up = -np.cos(incidence)

    return np.array([east, north, up])
```
