```python
import rasterio
import numpy as np

dir = '/Volumes/T9_InSAR/2026-04-14_nevada/S1_A014/insar/merged/'

input_vrt = dir + "topophase.flat.geo.vrt"
output_tif = dir + "topophase.flat.geo.tif"

with rasterio.open(input_vrt) as src:
    complex_data = src.read(1)

    wrapped_phase = np.angle(complex_data).astype(np.float32)

    # Build clean GeoTIFF metadata
    meta = {
        "driver": "GTiff",
        "height": src.height,
        "width": src.width,
        "count": 1,
        "dtype": "float32",
        "crs": src.crs,
        "transform": src.transform,
        "compress": "deflate"
    }

with rasterio.open(output_tif, "w", **meta) as dst:
    dst.write(wrapped_phase, 1)

print("Wrapped phase written to:", output_tif)
```
