```python
from datetime import datetime

def yyyydoy_to_date(yyyydoy):
    """Convert YYYYDOY (e.g., 2024302) to YYYY-MM-DD."""
    date = datetime.strptime(yyyydoy, "%Y%j")
    return date.strftime("%Y-%m-%d")

def date_to_yyyydoy(date_str):
    """Convert YYYY-MM-DD to YYYYDOY (e.g., 2024302)."""
    date = datetime.strptime(date_str, "%Y-%m-%d")
    return date.strftime("%Y%j")
```

```python
# Convert from YYYYDOY to YYYY-MM-DD
doy_str = "2024302"
date_str = yyyydoy_to_date(doy_str)
print(f"{doy_str} -> {date_str}")

```

```python
# Convert from YYYY-MM-DD to YYYYDOY
date_str = "2024-10-28"
yyyydoy = date_to_yyyydoy(date_str)
print(f"{date_str} -> {yyyydoy}")
```
